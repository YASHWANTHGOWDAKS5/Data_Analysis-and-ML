```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv("students_marks_dataset.csv")
```

```python
data.head()
```

```python
data.columns
```

```python
X = data[['Attendance(%)', 'CurrentMarks', 'ProjectScore']]
y = data['NextSemMarks']
X_train,X_test,y_train,y_test = train_test_split(
    X,y,
    random_state=42,
    test_size=0.2
)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
print("R2_score: ",round(r2_score(y_test,y_pred),2))
```

```python
new_student = pd.DataFrame([{
    "Attendance(%)": 87,
    "CurrentMarks": 88,
    "ProjectScore": 7.2
}])
print("New student data:\n ",new_student)
print("Next Sem predicted score is: ",round(model.predict(new_student)[0],2))
```
