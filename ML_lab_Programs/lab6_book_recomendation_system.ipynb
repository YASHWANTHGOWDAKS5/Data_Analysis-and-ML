```python
#history of books
history = {
    "s1":["Machine_learning","AI","Deep_learning"],
    "s2":["python","Data_science","AI"],
    "s3":["Novel","poetry","story"],
    "s4":["fiction","Machine_learning"],
    "s5":["Kannada","AI","English"],
    "s6":["Novel","fiction","story"],
}
```

```python
tech =["Machine_learning","AI","Deep_learning","python","Data_science"]
type={}
for books in history.values():
    for b in books:
        if b in tech:
            type[b]="Tech"
        else:
            type[b]="Literature"
type
```

```python
#concept learning
concept = {}
for books in history.values():
    cat=[type[b] for b in books]
    for c in cat:
        concept[c] = concept.get(c,0)+1
print(concept)
dominant = max(concept, key=concept.get)
print(dominant)
```

```python
#Logic for recomendation
def recommend_book(current_books):
    cur = [type[b] for b in current_books]
    if dominant in cur:
        return [b for b,t in type.items() if t == dominant and b not in current_books]
    elif dominant not in cur:
        return [b for b,t in type.items() if t != dominant and b not in current_books]
    else:
        return [b for b,t in type.items() if b not in current_books]
```

```python
#Lets check our recommendation system
print("recomendation for AI: ",recommend_book(["AI"]))
print("------------  ------------")
print("recomendation for fiction: ",recommend_book(["fiction"]))
```
