3. Performing Bivariat and univariat analyssis on food preference dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("food_preferences_survey.csv")
```

```python
data.head()
```

```python
#frequency of the food items
print("Food items count: ")
data["preferred_food"].value_counts()
```

```python
#Univariate analysis
sns.countplot(
    data=data,
    x='preferred_food',
    color='lightgreen'
)
plt.xlabel("Food items")
plt.ylabel("Count")
plt.title("Food item counts")
plt.show()
```

```python
exp=[0,0,0.1,0.1,0.1,0]
counts=data["preferred_food"].value_counts()
plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    shadow=True,
    explode=exp,
    startangle=90
)
plt.title("pie chart for food counts")
plt.show()
```

```python
# Bivariate analysis
# gender Vs Preferred Food
print(pd.crosstab(data.gender,data.preferred_food))
sns.countplot(
    data=data,
    x='preferred_food',
    hue='gender'
)
plt.title("Gender Vs Prefered food")
plt.show()
```

```python
# Bivariate analysis
# year_of_study Vs Preferred Food
print(pd.crosstab(data.year_of_study,data.preferred_food))
sns.countplot(
    data=data,
    x='preferred_food',
    hue='year_of_study'
)
plt.title("Year of Study Vs Prefered food")
plt.show()
```
