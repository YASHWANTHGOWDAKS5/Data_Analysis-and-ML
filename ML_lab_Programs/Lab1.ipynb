1.Perform eda on student dataset to identify key factors influencing student exam score

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("lab1.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
#Scatter plot to see the relation of exam_score with remaining
plt.figure(figsize=(5,4))
plt.scatter(
    data=df,
    x='exam_score',
    y='attendence',
    color='blue'
)
plt.xlabel("Exam_score")
plt.ylabel("Attendence")
plt.title("Scatter plot between Exam Score Vs Attendence")
plt.show()
```

```python
plt.figure(figsize=(5,4))
plt.scatter(
    data=df,
    x='study_hours',
    y='exam_score',
    color='green'
)
plt.xlabel("Study hour")
plt.ylabel("Exam_score")
plt.title("Scatter plot between study_hours Vs Exam Score")
plt.show()
```

```python
plt.figure(figsize=(5,4))
plt.scatter(
    data=df,
    x='exam_score',
    y='class_participation',
    color='orange'
)
plt.xlabel("Exam Score")
plt.ylabel("Class Participation")
plt.title("Scatter plot between exam_score Vs Class Participation")
plt.show()
```

```python
#Lets check interquartile range
stats = df.describe()
IQR_study_hour = stats.loc['75%','study_hours']-stats.loc['25%','study_hours']
IQR_class_participation = stats.loc['75%','class_participation']-stats.loc['25%','class_participation']
IQR_attendence = stats.loc['75%','attendence']-stats.loc['25%','attendence']
IQR_exam_scores = stats.loc['75%','exam_score']-stats.loc['25%','exam_score']
print(f"IQR_study_hour: {IQR_study_hour}\nIQR_class_participation: {IQR_class_participation}\nIQR_attendence: {IQR_attendence}\nIQR_exam_scores: {IQR_exam_scores}")
```
