```python
data=df[['volatile acidity']]
Data_4=df[['fixed acidity', 'citric acid', 'residual sugar',
       'chlorides']]
```

```python
q25,q50,q75=np.percentile(data,[25,50,75])
iqr=q75-q25

min=q25-1.5*(iqr)
max=q75+1.5*(iqr)
maximum=max
print(f'min = {min} \n q25 = {q25}\n q50 = {q50}\n q75 = {q75}\n max = {max}')
print(type(maximum))
```

```python
a=[]
b=[]
print('Number of values in Volatile acidity column:',len(df['volatile acidity']))
for i,x in enumerate(df['volatile acidity']):
    if float(x) > maximum or float(x) < min:
        a.append(f'Pos:{i} -> {x}')
        b.append(x)
print('Number of outliers:',len(a))
for e in a:
    print(e)
```

```python
sns.histplot(b,bins=40)
```

```python
df_cleaned = df[(df['volatile acidity'] >= min) & (df['volatile acidity'] <= maximum)]

print(f"Original number of rows: {len(df)}")
print(f"Number of rows after removing outliers: {len(df_cleaned)}")
```
